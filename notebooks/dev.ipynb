```python
def fetch_open_targets_disease_data(disease_id):
    """
    Fetch disease-drug associations from Open Targets for a given disease using the GraphQL API.
    Args:
    - disease_id (str): Open Targets ID for the disease (e.g., EFO_0000519 for glioblastoma).

    Returns:
    - JSON response containing the associated targets.
    """
    url = "https://api.platform.opentargets.org/api/v4/graphql"
    
    query = """
    query ($diseaseId: String!) {
      disease(efoId: $diseaseId) {
        name
        associatedTargets {
          rows {
            target {
              approvedSymbol
            }
          }
        }
      }
    }
    """
    
    variables = {"diseaseId": disease_id}
    
    response = requests.post(url, json={"query": query, "variables": variables})
    
    if response.status_code == 200:
        return response.json()
    else:
        return {"error": f"Failed to fetch data for disease {disease_id}. Status code: {response.status_code}, Response: {response.text}"}

# Test with the correct EFO ID for glioblastoma
glioblastoma_data = fetch_open_targets_disease_data("EFO_0000519")
print(glioblastoma_data)
```
